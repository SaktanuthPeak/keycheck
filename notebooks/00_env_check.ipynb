### Resume test (YOLO11n สั้นๆ)
ตั้ง `RUN_RESUME_TEST=True` เพื่อฝึก 2 epoch, หยุดหลัง epoch 1 (รันเซลล์นี้ 2 ครั้ง: ครั้งแรกขัดด้วย Interrupt kernel หลัง epoch 1 หรือปล่อยให้จบ) — ใช้ข้อมูลจาก smoke run ของ Q2

```python
RUN_RESUME_TEST = False
SMOKE_DATA = "outputs/pipeline/qwertz_v1_smoke/q2_rectified/rectified_ppu96/data.yaml"
if RUN_RESUME_TEST:
    from ultralytics import YOLO
    project = Path("outputs/env_check").resolve()     # Ultralytics nests a relative project under runs/<task>/
    last = project / "train" / "weights" / "last.pt"
    if last.exists():
        YOLO(str(last)).train(resume=True)
    else:
        YOLO("yolo11n.pt").train(data=SMOKE_DATA, epochs=2, imgsz=640, batch=8, project=str(project), name="train", exist_ok=True,
                                 fliplr=0, flipud=0, mosaic=0, mixup=0)
```
